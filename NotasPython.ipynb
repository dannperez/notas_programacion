```python
lista_int = [1,2,3,4]
lista_cad = ["hola","mundo"]
lista_comb = ["hola", 1, 2, True,"platanos"]

matriz = [[1,2,3],
          [4,5,6],
          [7,8,9]]   # Lista de listas
print(matriz)

# Para acceder a elementos por indice
print("Para acceder a elementos por indice: ")
print(lista_comb[0])
print(lista_comb[-1]) # Cuenta desde el final

#Para cceder a los elementos de una matriz
print("Acceder a los elementos de la matriz")
print(matriz[1][0]) # Renglon 1 columna 0
```

```python
# Slicing

lista1 = [1,2,3,4,5,6]
print(lista1[1:4]) # No toma la posicion 4
print(lista1[:3])  # Devuelve los tres primeros numeros
print(lista1[3:])  # Devuelve los ultimos 2 numeros

# Para hacer una copia de la lista
print(lista1[:])

# Omitir valores
print(lista1[::2]) # Indices pares
print(lista1[::3]) # Indices impares

# Revertir lista
print(lista1[::-1])

# Modificar listas
lista1[0] = 20
print(lista1)

# Agregar elementos a una lsita
lista2 = [1,2,3,4,5]
lista2 += [7,8,9]
print(lista2)

#Recuperar longitud de una lista
len(lista2)
```

```python
# Methods
#Agregar elementos en las listas

lista1 = [1,2,3,4,5,6]
lista1.append(7) # Agregar elementos al final de la lista

lista2 = ['a','b','c','d']
lista2.insert(1,'k') # Agregar elementos en una posicion especifica

lista2.extend(['m','g','d']) # Agregar varios elementos al final de la lista

print(lista1)
print(lista2)
```

```python
# Eliminar elementos

lista2 = ['a','k','b','c','d','m','g','d']
lista2.remove('d') # Borra solo el primer elemento 'd'
print(lista2)

#Por indice
lista2 = ['a','k','b','c','d','m','g','d']
lista2.pop() # Elimina el ultimo elemento de la lista
print(lista2)

lista2 = ['a','k','b','c','d','m','g','d']
lista2.pop(1) # Elimina el elemento en posicion 1
print(lista2)

lista2 = ['a','k','b','c','d','m','g','d']
del lista2[2:5] # Elimina un rango de la lista
print(lista2)

lista2 = ['a','k','b','c','d','m','g','d']
lista2.clear() # Elimina todo slos elementos de la lista
print(lista2)
```

## Bucle While y break
Permiten ejecutar un bloque de codigo repetidamente mientras se cumple una condicion
****
### Palabras clave
* **break**: Rompe el bucle. Lo usamos cuando no sabemos la condicion de salida
* **continue***: Salta la iteracion en concreto y continua con la siguiente iteracion ignorando el codigo siguiente
* **else**: entra cuando el bucle termina
* **try ... except**: para evitar que se muestre un error en la consola cuando el usuario no introduce lo esperado

```python
contador = 0
while contador < 5:
  print(contador)
  contador += 1  # Importante actualizarlo
```

```python
contador = 0
while contador <= 100:
  contador += 1
  print(contador)
  if contador % 5 == 0:
    print("El numero es multiplo de 5")
    break
```

```python
contador = 0
while contador <= 10:
  contador += 1
  if contador % 2 == 0: # Es multiplo de 2
    continue
  print(contador)
```

```python
# Pedir al usuario un numero dentro de un while
numero = -1
while numero < 0:
  numero = int(input("Introduce un numero positivo: "))
  if numero < 0:
    print("EL numero debe ser positivo.")

print(f"EL numero es {numero}")
```

```python
# Try ...  except
numero = -1
while numero < 0:
  try:
    numero = int(input("Introduce un numero positivo: "))
    if numero < 0:
      print("El numero debe ser positivo.")
  except:
    print("Debes introducir un numero entero") # Evita mostrar el error cuando no se introducen enteros

print(f"EL numero es {numero}")
```

## For
Ejecutar un bloque de codigo repetidamente mientras itera sobre algo

Las cadenas son iterables

* **enumerate( )**: devuelve los indices de los elementos (siempre devuelve primero el indice y luego el valor)
* **range( )*: crea una secuencia de numeros (range(inicio, fin, paso))

```python
nombre = "Danna"
for letra in nombre:
  print(letra)

frutas = ['manzana', 'mango', 'durazno']

for index, fruta in enumerate(frutas):
  print(f"{fruta} esta en el indice {index}")
```

```python
# For anidados

letras = ['A','B','C']
num = [1,2,3]

for i in letras:
  for j in num:
    print(f"{i}{j}")


# Break
animales = ["perro",'gato','pez','loro']
for animal in animales:
  print(animal)
  if animal == "pez":
    break

print("-------------")

#Continue
animales = ["perro",'gato','pez','loro']
for animal in animales:
  if animal == "pez":
    continue
  print(animal)
```

```python
#Comprension de listas
animales = ["perro",'gato','pez','loro']
animales_mayus =[animal.upper() for animal in animales]
print(animales_mayus)

pares = [num for num in [1,2,3,4,5,6] if num % 2 == 0]
print(pares)
```

```python
nums = range(5) # NO es una lista

# range(inicio, fin, paso)
for num in range(0,10,2):
  print(num)

# Crear listas a partir de range()
for _ in range(3):
  nums = range(5)
  lista = list(nums)
  print(lista)
```

## Diccionarios
Evitan los for anidados

* **diccionario.pop("valor")** : para eliminar un valor y mostrarlo en consola
* **dicc_a.update(dicc_b)**: para actualizar la info del dicc_a con la info del dicc_b
* **print("valor" in "diccionario")**: ver si la prop "valor" esta en "diccionario"
* **print("diccionario".keys())**: obtener las claves
* **print("diccionario".values())**: obtener los valores


```python
persona = {
  "nombre": "Danna",
  "edad" : 20,
  "socials":{
      "fb" : "@fb",
      "ig" : "@ig"

  },
  "mascotas" : "molly"
}

# Acceder a los valores
print(persona["nombre"])
print(persona["socials"]["fb"])

# Cambiar valores
persona["nombre"] = "Paola"

# Eliminar una propiedad
del persona["edad"]

# Imprime diccionario
print(persona)

# ELiminar y acceder al valor con persona.pop("valor")
```

```python
#Sobreescribir un diccionario
b = {"nombre": "brandon", "edad":20, "mascotas":"cofis"}

persona.update(b)
print(persona)

#Obtener claves y valores
print("\nkeys")
print(persona.keys())
print("\nvalues")
print(persona.values())
```

## Expresiones regulares
Secuencia de caracteres que forman un patron de busqueda.

Se utilizan para la busqueda de cadenas de texto, validacion de datos etc.

****

Regex
* Busqueda avanzada: enocntrar patrones en cadenas de texto

* Validacion de datos: asegurar que los datos ingresados por el usuaro sean correctos.

* Manipulacion del texto: modificar, manipular, extraer y reemplazar partes de una cadena de texto


****

```python
import re
```

```python
# Creamos patron
pattern = "Hola"
# Texto donde queremos buscar
text = "Hola mundo"
# Funcion re
result = re.search(pattern, text)

if result:
  print("He encontrado el patron en el texto")
else:
   print("No he encontrado el patron en el texto")
```

****
###Funciones

```python
# Creamos patron
pattern = "Python"
# Texto donde queremos buscar
text = "Me gusta Python, porque Python es bueno y Python ..."

# .group() devuelve la cadena que coincide con el patron
print(result.group())

# .start() devuelve la posicion incial de la coincidencia
print(result.start())

# .end() devuelve la posicion final de la coincidencia
print(result.end())

# re.findall(pattern, text) devuelve lista con todas las coincidencias
matches = re.findall(pattern, text)
print(matches)
print(len(matches))

#Sustituir una letra con un punto para encontrar un patron con cualquier letra ahi
print("_________________________________________________________")
pattern = "Pyt.on"
# Texto donde queremos buscar
text = "Me gusta Pytton, porque Pytcon es bueno y Python ..."
matches = re.findall(pattern, text)
print(matches)
```

## Fechas

```python
from datetime import datetime, timedelta
import locale
# Ver documentacion de python para los formatos de fechas
```

```python
#Cambiar nombreas a spanish
locale.setlocale(locale.LC_TIME, '')


now = datetime.now()
print(f"Fecha y hora actual {now}")

specific_date = datetime(2004,9,4,10,30)
print(f"\nFecha especifica: {specific_date}")

# Formatear fechas
format_date= now.strftime("%d/%m/%Y")
print(f"\nLa fecha actual es: {format_date}")

# Restar fechas
yesterday = datetime.now() - timedelta(days=1)
print(f"\nAyer fue: {yesterday}")

atras = datetime.now() - timedelta(hours=1)
print(f"\nHace una hora: {atras}")

cumple = datetime(2025,9,4)
hoy = datetime.now()
print(f"\nFaltan {cumple-hoy} para mi cumpleanios")
```
